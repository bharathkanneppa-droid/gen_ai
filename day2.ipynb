17.1 Concept Checks - Q&A

1. Define AI, ML, and DL, and explain how they are nested.
AI is the big goal - making machines act smart. ML is one way to do that, where the machine learns patterns from data instead of being told exact rules. DL is a type of ML that uses layered neural networks to learn even more complex patterns. So it's like AI > ML > DL, each one sitting inside the bigger one.

2. Give one everyday example each of classification, regression, and clustering.
Classification: your email app sorting mail into "spam" or "not spam." Regression: predicting tomorrow's temperature based on past weather data. Clustering: Netflix grouping users with similar watching habits without knowing their names in advance.

3. Why is all current AI considered "Narrow" (ANI)?
Because every AI system today is trained to do one specific job well, like translating text or recognizing faces, and it can't just switch to a totally different task on its own. True general intelligence (AGI), where a machine can think and adapt like a human across any domain, doesn't exist yet.

4. Explain the difference between supervised and unsupervised learning with your own example.
In supervised learning, I give the model labeled data, like photos already tagged "cat" or "dog," so it learns to predict labels for new photos. In unsupervised learning, I just give raw data with no labels, like customer purchase records, and the model finds hidden groups or patterns on its own.

5. What is the difference between Generative AI and Predictive AI?
Predictive AI looks at existing data and predicts an outcome, like whether a loan applicant will default. Generative AI creates something new, like writing text, generating images, or composing code, based on patterns it learned.

6. Why is data quality more important than data quantity?
Because a model learns exactly what you feed it - if the data is messy, biased, or wrong, having millions of bad examples just teaches the model to be confidently wrong. A smaller set of clean, accurate data usually gives better real-world results than a huge pile of dirty data.

7. List the 7 stages of the AI project lifecycle in order.
1) Problem definition, 2) Data collection, 3) Data preparation/cleaning, 4) Model selection, 5) Model training, 6) Model evaluation/testing, 7) Deployment and monitoring.

8. Explain "Garbage In, Garbage Out" in the context of bias.
If the training data has biased or unfair patterns, like historical hiring data that favored one group over another, the model will learn and repeat that same bias in its predictions. The model doesn't know right from wrong - it just reflects whatever "garbage" (bias, errors, gaps) was in the data it was fed.

Which of the following is the correct nesting order from broadest to narrowest?
A) ML → AI → DL
B) DL → ML → AI
C) AI → ML → DL
D) AI → DL → ML

Correct Answer: C) AI → ML → DL — AI is the overall goal of making machines intelligent, ML is a subset of AI that learns from data, and DL is a subset of ML using layered neural networks.

Question 2 (Multiple Choice)
A model is trained to sort emails into "Spam" or "Not Spam." Which type of ML task is this?
A) Regression
B) Clustering
C) Classification
D) Reinforcement Learning

Correct Answer: C) Classification — because the model is predicting a discrete category (label) for each input, rather than a continuous number (regression) or grouping unlabeled data (clustering).
