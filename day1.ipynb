```python
def is_prime(num):
    if num<2:
        return False
    else:
        for i in range(2,num//2):
            if num%i==0:
                return False
    return True
n=int(input("Enter the number : "))
print(is_prime(n))
#factorial of number
def fact(n):
    if n==1:
        return 1
    return n*fact(n-1)
n=int(input("enter the number : "))
print(fact(n))
#max of three numbers 
def max_of_three(a,b,c):
    if a>=b and a>=c:
        return a
    elif b>=c:
        return b
    return c
a=int(input("enter the number : "))
b=int(input("enter the number : "))
c=int(input("enter the number : "))
print(max_of_three(a,b,c))
```

```python
arr=[4,2,8,6,2,8,4]
def unique_elements(arr):
    unique=[]
    for i in arr:
        if i in unique:
            unique.append(i)
    return unique
print(unique_elements(arr))
```
